```python
import pandas as pd
import numpy as np

# 1. Load the data
df = pd.read_excel(r"C:\Users\PRIYDARSHINI\Documents\data_collection.xlsx")

# 2. Clean: Remove commas and convert to numeric
cols_to_clean = ['GHG Emission Intensity', 'Water Intensity', 'Energy Consumption', 
                 'Waste Generated', 'Female Representation']
for col in cols_to_clean:
    df[col] = pd.to_numeric(df[col].astype(str).str.replace(',', ''), errors='coerce')

# 3. Impute missing values with column mean
numeric_cols = df.select_dtypes(include=[np.number]).columns
df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].mean())

# 4. Define Directionality
lower_is_better = ['GHG Emission Intensity', 'Water Intensity', 'Energy Consumption', 
                   'Energy Intensity', 'Waste Generated', 'Worker LTIFR']
higher_is_better = ['Female Representation', 'Average Training Hours', 'CSR Expenditure', 
                    'Independent Directors', 'Board Governance']

# 5. Perform Normalization
for col in numeric_cols:
    min_val = df[col].min()
    max_val = df[col].max()
    
    if col in lower_is_better:
        # Invert: Higher values get lower scores
        df[col] = 1 - (df[col] - min_val) / (max_val - min_val)
    else:
        # Standard: Higher values get higher scores
        df[col] = (df[col] - min_val) / (max_val - min_val)

# 6. Save result
df.to_excel(r"C:\Users\PRIYDARSHINI\Documents\normalized_data_final.xlsx",index=False)
```
